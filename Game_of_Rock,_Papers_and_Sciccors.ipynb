```python
import random
def get_choices():
    player_choice = input("Enter your choice (rock, papers, sciccors: ")
    options = ["rock", "paper", "sciccors"]
    computer_choice = random.choice(options)
    choices = {"player": player_choice, "computer": computer_choice}
    return choices

def check_win(player, computer):
    print(f"You chose {player}, computer chose {computer}")
    if player == computer:
        return "Its a tie!"
    elif player == "rock":
        if computer == "scissors":
            return "Rock wins"
        else:
            return "Rock loses"
    elif player == "paper":
        if computer == "rock":
            return "Paper wins"
        else:
            return "Paper loses"
    elif player == "sciccors":
        if computer == "paper":
            return "sciccors wins"
        else:
            return "sciccors loses"

choices = get_choices()
result = check_win(choices["player"], choices["computer"])
print(result)
```
