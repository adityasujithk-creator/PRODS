# K Means Clustering

K-Means Clustering is an unsupervised machine learning algorithm used to group similar data points into K distinct clusters based on their features. The algorithm partitions the dataset such that data points within the same cluster are more similar to each other than to those in other clusters.

The center of each cluster is called a Centroid, and the algorithm iteratively updates these centroids to minimize the distance between data points and their assigned cluster center.

Centroid = 1/n​∑​xi​

J = ∑K ​∑Cj ​​∣∣xi​−μj​∣∣^2

Where:

K = Number of clusters

Cj= Cluster j

xi= Data point

μj= Centroid of cluster j

The goal is to make points as close as possible to their centroid.

K-Means Clustering is an unsupervised machine learning algorithm that partitions data into K clusters by assigning points to the nearest centroid and iteratively minimizing the within-cluster variance.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_blobs
```

```python
data = make_blobs(n_samples=200, n_features=2, centers=4, cluster_std=1.8, random_state=101)
```

```python
plt.scatter(data[0][:,0],data[0][:,1],c = data[1],cmap='rainbow')
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4)
```

```python
kmeans.fit(data[0])
```

```python
kmeans.cluster_centers_
```

```python
kmeans.labels_
```

```python
fig , (ax1,ax2) = plt.subplots(1,2, sharey=True,figsize=(10,6))

ax1.set_title('K_Means')
ax1.scatter(data[0][:,0],data[0][:,1],c = kmeans.labels_,cmap='rainbow')

ax2.set_title('Original')
ax2.scatter(data[0][:,0],data[0][:,1],c = data[1],cmap='rainbow')
```

It is very important to note: we actually have the labels for this data set, but we will NOT use them for the KMeans clustering algorithm, since it is an unsupervised learning algorithm.

When using the KMeans algorithm under normal circumstances, it is because you don't have labels. In this case, we will use the labels to try to get an idea of how well the algorithm performed, but you won't usually do this for KMeans, so the classification report and confusion matrix at the end of this project don't truly make sense in a real-world setting!

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=2)
```

```python
kmeans.fit(df.drop('Private',axis=1))
```

```python
def converter(private):
  if private == 'Yes':
    return 1
  else:
    return 0
```

```python
df['Cluster'] = df['Private'].apply(converter)
```

```python
df.head()
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(df['Cluster'],kmeans.labels_))
print('\n')
print(confusion_matrix(df['Cluster'],kmeans.labels_))
```
