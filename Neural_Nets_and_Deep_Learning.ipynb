# Perceptron Model

The Perceptron Model is the simplest type of Artificial Neural Network (ANN) and a fundamental building block of deep learning. It is a supervised learning algorithm used for binary classification, where the output belongs to one of two classes (e.g., Yes/No, 0/1, True/False).


A Perceptron is the simplest artificial neural network that performs binary classification by computing a weighted sum of inputs, adding a bias, and applying an activation function to produce an output.

A perceptron takes multiple inputs, applies weights to them, adds a bias, and passes the result through an activation function to produce an output.

Step 1: Calculate Weighted Sum:

z=∑ ​wi​xi ​+ b

where,

xi = Input features : Inputs are the features or data values provided to the perceptron/neural network for processing.

wi = Weights : Weights tell the neural network which inputs are more important.

b = Bias : Bias is an additional parameter added to the weighted sum that allows the model to shift the decision boundary and make better predictions.

Step 2: Apply Activation Function :

After calculating the weighted sum, the activation function decides the final output.

It introduces non-linearity, allowing neural networks to learn complex patterns.

# Neural Networks(NN)

A Neural Network is a machine learning model composed of interconnected artificial neurons arranged in layers that learn patterns from data by adjusting weights and biases to make predictions or decisions.

They are widely used for tasks such as image recognition, speech recognition, natural language processing, recommendation systems, and autonomous systems.

# Cost Functions

A Cost Function (Loss Function) is a mathematical function that measures the difference between the actual output and the predicted output of a machine learning model.

The goal of training is to minimize the cost function to improve model performance.

Types:

1. Mean Squared Error:

    MSE = 1/N​∑​(yi​−y^​i​)^2

    Where:

    N = Number of samples

    yi = Actual value

    y^i(y cap i) = Predicted value

2. Mean Absolute Error (MAE):
Uses absolute differences.

    MAE = 1/N ∑ ∣yi − y^i|

3. Root Mean Squared Error (RMSE):
Square root of MSE.

    RMSE = 1/N ∑ (yi − y^i)**2

    Provides error in the original units.

4. Binary Cross-Entropy (Log Loss):
Used in Logistic Regression and Binary Classification.

    J = −1/m ∑ [yi log(y^i)+(1−yi)log(1− y^i)]

    Where:

    yi= Actual label

    y^i(y cap i) = Predicted probability

# Gradient Descent

Gradient Descent is an optimization algorithm used to minimize the cost function by iteratively adjusting the model's weights and biases in the direction of the steepest decrease in error.

Step 1

Initialize weights randomly.

Step 2

Calculate prediction.

Step 4

Calculate gradient.

Gradient indicates: Which direction reduces error?

Step 5

Update weights.

Weight Update Rule:

w new = w old −η(∂J/∂w)

Where:

w = Weight

J = Cost Function

η = Learning Rate : Learning Rate determines the size of each step.

∂J/∂w = Gradient

Types:

1. Batch Gradient Descent

    Uses the entire dataset for each update.

2. Stochastic Gradient Descent (SGD)

    Uses one training example at a time.

3. Mini-Batch Gradient Descent

    Uses small batches.

# Back Propagation

Backpropagation (Backward Propagation of Errors) is a supervised learning algorithm used to train neural networks by calculating the error at the output layer and propagating it backward through the network to update weights and biases.

It works together with Gradient Descent to minimize the loss (cost) function and improve the model's predictions.

Without Backpropagation:

Neural networks cannot learn.
Weights remain unchanged.
Prediction accuracy does not improve.

*   Neural networks cannot learn.
*   Weights remain unchanged.


*   Prediction accuracy does not improve.





Step 1: Forward Propagation

The input passes through the network.

Apply activation: The network produces a prediction.

Step 2: Calculate Loss

Compare prediction with actual value.

Loss is computed using a cost function.

Step 3: Backward Propagation

The error is propagated backward.

The network calculates: How much did each weight contribute to the error?

Step 4: Compute Gradients

Backpropagation computes: ∂L/∂w

Chain Rule in Backpropagation:

Backpropagation relies on the Chain Rule of Calculus.

For example:

L=f(a) and

a=g(z) then

dL/dz = dL/da × da/dz

The chain rule allows gradients to be calculated layer by layer from the output back to the input.

Step 5: Update Weights

After computing gradients:

Weights are updated using Gradient Descent.

If a weight contributes heavily to the error, it is adjusted more.

Backpropagation is a neural network training algorithm that calculates gradients of the loss function using the chain rule and propagates errors backward through the network to update weights and biases, thereby minimizing prediction error.
