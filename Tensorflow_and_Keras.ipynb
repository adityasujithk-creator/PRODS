# Tensorflow and Keras

TensorFlow is an open-source Machine Learning and Deep Learning framework developed by Google for building, training, and deploying machine learning models. It provides tools and libraries for numerical computation, neural networks, deep learning, and artificial intelligence applications.

TensorFlow allows developers and researchers to create models that can run on CPUs, GPUs, TPUs, mobile devices, and cloud platforms.

Main Components of TensorFlow
1. Tensors

    A tensor is a multi-dimensional array used to store data.

2. Operations (Ops)

    Operations perform computations on tensors.

3. Computational Graph

    TensorFlow represents computations as graphs.

4. Variables

    Variables store trainable parameters.

    During training:

    Weights change & Biases change

    to improve predictions.

TensorFlow Workflow

Step 1: Load Data

Step 2: Build Model

Step 3: Train Model

Step 4: Evaluate Model

Step 5: Predict

TensorFlow is an open-source machine learning and deep learning framework developed by Google that uses tensors and computational graphs to build, train, and deploy AI models efficiently.

Keras is a high-level open-source Deep Learning API that allows developers to build, train, and deploy neural networks easily. It provides a simple and user-friendly interface for creating deep learning models and runs on top of frameworks such as TensorFlow.

Today, Keras is the official high-level API integrated into Google's TensorFlow and is accessed through tf.keras.

Keras Workflow

Data

 ↓

Build Model

 ↓

Compile Model

 ↓

Train Model

 ↓

Evaluate Model

 ↓

Predict

Main Components of Keras
1. Layers

    Layers are the building blocks of neural networks.

2. Model

    A model is a collection of layers.

3. Loss Function

    Measures prediction error.

4. Optimizer

    Updates weights using Gradient Descent.

5. Metrics

    Evaluate performance.

Keras is a high-level deep learning API integrated with TensorFlow that provides a simple and efficient way to build, train, evaluate, and deploy neural network models.

TF Regression - Data Preprocessing and Model Creation

```python
X = df.drop('price',axis=1).values
y = df['price'].values
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
x_train = scaler.fit_transform(X_train)
x_test = scaler.transform(X_test)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
x_train.shape
```

```python
model = Sequential()

model.add(Dense(19,activation='relu'))
model.add(Dense(19,activation='relu'))
model.add(Dense(19,activation='relu'))

model.add(Dense(1))

model.compile(optimizer='adam',loss='mse')
```

```python
model.fit(x=x_train,y=y_train,validation_data=(x_test,y_test),batch_size=128,epochs=400)
```

```python
model_loss = pd.DataFrame(model.history.history)
```

```python
model_loss.plot()
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,explained_variance_score
```

```python
predictions = model.predict(x_test)
```

```python
mean_absolute_error(y_test,predictions)
```

```python
np.sqrt(mean_squared_error(y_test,predictions))
```

```python
df['price'].describe()
```

```python
explained_variance_score(y_test,predictions)
```

```python
plt.figure(figsize=(12,8))
plt.scatter(y_test,predictions)
plt.plot(y_test,y_test,'r')
```

```python
x = df.drop('benign_0__mal_1',axis=1).values
y = df['benign_0__mal_1'].values
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=101)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
x_train = scaler.fit_transform(x_train)
```

```python
x_test = scaler.transform(x_test)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
```

```python
x_train.shape
```

```python
model = Sequential()

model.add(Dense(30,activation='relu'))
model.add(Dropout(0.5))

model.add(Dense(15,activation='relu'))
model.add(Dropout(0.5))

model.add(Dense(1,activation='sigmoid'))

model.compile(loss='binary_crossentropy',optimizer='adam')

model.add
```

```python
model.fit(x=x_train,y=y_train,epochs=600,validation_data=(x_test,y_test))
```

```python
losses = pd.DataFrame(model.history.history)
```

```python
losses.plot()
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
early_stop = EarlyStopping(monitor='val_loss',mode='min',verbose=1,patience=25)
```

```python
model.fit(x=x_train,y=y_train,epochs=600,validation_data=(x_test,y_test),callbacks=[early_stop])
```

```python
model_loss = pd.DataFrame(model.history.history)
model_loss.plot()
```

```python
from tensorflow.keras.layers import Dropout
```

```python
model = Sequential()

model.add(Dense(30,activation='relu'))
model.add(Dropout(0.5))

model.add(Dense(15,activation='relu'))
model.add(Dropout(0.5))

model.add(Dense(1,activation='sigmoid'))

model.compile(loss='binary_crossentropy',optimizer='adam')
```

```python
model.fit(x=x_train,y=y_train,epochs=600,validation_data=(x_test,y_test),callbacks=[early_stop])
```

```python
model_loss = pd.DataFrame(model.history.history)
model_loss.plot()
```

```python
pred_probs = model.predict(x_test)

predictions = (pred_probs > 0.5).astype(int)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
print(classification_report(y_test,predictions))
print(confusion_matrix(y_test,predictions))
```
